```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings('ignore')
```

```python
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_circles
X, _=make_circles(n_samples=800, noise=0.1, factor=0.3)
```

```python
X
```

```python
plt.scatter(X[:,0],X[:,1])
```

```python
scan=DBSCAN(eps=0.1)
scan.fit_predict(X)
```

```python
plt.scatter(X[:,0],X[:,1], c=scan.labels_)
```
