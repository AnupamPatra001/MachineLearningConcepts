```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# reading data from scikit learn
from sklearn.datasets import load_diabetes
dia=load_diabetes()
dia
```

```python
data=pd.DataFrame(dia.data, columns=dia.feature_names)
data
```

```python
data['target']=dia.target
```

```python
data
```

```python
data.isnull().sum()
```

```python
corr=data.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
```

```python
X=data[['bmi']]
y=data['target']
```

```python
X
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)
# random state splits the data in random manner in given times
```

```python
X_train
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model
```

```python
model.fit(X_train, y_train)
```

```python
model.coef_
```

```python
model.intercept_
```

```python
ypred=model.predict(X_test)
```

```python
ypred
```

```python
y_test
```

```python
plt.scatter(X_test, y_test, color='black', label='Actual')
plt.plot(X_test, ypred, color='blue', label='Actual')
plt.xlabel('BMI')
plt.ylabel('One year progress')
plt.legend()
plt.show()
```

```python
# evaluation
from sklearn.metrics import mean_squared_error as mse
mse(y_test, ypred)
```
