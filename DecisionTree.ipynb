```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')
```

```python
from sklearn.datasets import load_iris
data=load_iris()
data
```

```python
df=pd.DataFrame(data.data, columns=data.feature_names)
df
```

```python
X=df
y=data.target
```

```python
y
```

```python
X
```

```python
from sklearn.model_selection import train_test_split as tts
X_train, X_test, y_train, y_test = tts(X, y, test_size=0.3, random_state=1)
```

```python
X_train.shape, X_test.shape
```

```python
# decision tree
from sklearn.tree import DecisionTreeClassifier
classifier = DecisionTreeClassifier(criterion='entropy')
classifier
```

```python
classifier.fit(X_train, y_train)
```

```python
# how to visualize the tree
from sklearn import tree
plt.figure(figsize=(12,10))
tree.plot_tree(classifier, filled= True)
plt.show()
```

```python
y_pred=classifier.predict(X_test)
y_pred
```

```python
# classifier.predict_proba(X_test)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix
accuracy_score(y_test, y_pred)
```

```python
confusion_matrix(y_test, y_pred)
```

```python
# to remove overfitting >> post pruning > cut after complete tree generation
# its good for small data set

# observe the dt plot, the last split is actually overfitting , deviding the data in 
# such a way that each node is 100 percent pure leading to low number of dps
```

```python
# decision tree
# from sklearn.tree import DecisionTreeClassifier
classifier = DecisionTreeClassifier(criterion='entropy', max_depth=2)
classifier
classifier.fit(X_train, y_train)
# how to visualize the tree
# from sklearn import tree
plt.figure(figsize=(12,10))
tree.plot_tree(classifier, filled= True)
plt.show()
```

```python
y_pred=classifier.predict(X_test)
y_pred
```

```python
accuracy_score(y_test, y_pred)
```

```python
# as we can see the accuracy score is same as before, the more depth makes the tree overfitted
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred))
```

```python
# prepruning using hyper parameter tuning
param={
    'criterion':['gini', 'entropy', 'log_loss'],
    'splitter':['best','random'],
    'max_depth':[1,2,3,4,5,6,7,10],
    'max_features':['sqrt','log2','auto']
}

from sklearn.model_selection import GridSearchCV
model=GridSearchCV(DecisionTreeClassifier(), param_grid=param, cv=5, scoring='accuracy', verbose=2)
model
```

```python
model.fit(X_train, y_train)
```

```python
model.best_params_
```

```python
model.best_score_
```

```python
model.best_estimator_
```

```python
y_pred=model.best_estimator_.predict(X_test)
y_pred
```

```python
accuracy_score(y_test, y_pred)
```

```python
plt.figure(figsize=(12,10))
tree.plot_tree(model.best_estimator_, filled= True)
plt.show()
```

```python
confusion_matrix(y_test, y_pred)
```
