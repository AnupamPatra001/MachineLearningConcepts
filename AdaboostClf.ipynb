```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split

from sklearn.ensemble import AdaBoostClassifier
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score

import warnings
warnings.filterwarnings('ignore')
```

```python
X, y= make_classification(n_samples=1000, n_features=20, n_classes=2, random_state=1)
```

```python
X
```

```python
y
```

```python
X_train, X_test, y_train, y_test= train_test_split(X, y, test_size=0.3, random_state=1)
```

```python
X_train.shape, y_test.shape
```

```python
clf=AdaBoostClassifier()
clf
```

```python
clf.fit(X_train, y_train)
```

```python
y_pred=clf.predict(X_test)
```

```python
accuracy_score(y_test, y_pred)
```

```python
from sklearn.model_selection import GridSearchCV
params={
    'n_estimators':[50, 100, 200],
    'learning_rate':[0.01, 0.1, 1, 1.5, 2, 10],
    'algorithm':['SAMME']
}
```

```python
grid=GridSearchCV(AdaBoostClassifier(),param_grid=params, cv=5, verbose=2, scoring='accuracy')
```

```python
grid
```

```python
grid.fit(X_train, y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred=grid.best_estimator_.predict(X_test)
```

```python
accuracy_score(y_test, y_pred)
```

```python
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split

from sklearn.ensemble import AdaBoostRegressor
from sklearn.metrics import r2_score

import warnings
warnings.filterwarnings('ignore')
```

```python
X, y=make_regression(n_samples=1000, n_features=2, noise=10, random_state=1)
```

```python
X_train, X_test, y_train, y_test= train_test_split(X, y, test_size=0.3, random_state=1)
```

```python
X_train.shape, y_test.shape
```

```python
params={
    'n_estimators':[50, 100, 200],
    'learning_rate':[0.01, 0.1, 1, 1.5, 2, 10],
    'loss':['exponential', 'linear', 'square']
}
grid2=GridSearchCV(AdaBoostRegressor(), param_grid=params, cv=5, verbose=2)
```

```python
grid2.fit(X_train, y_train)
```

```python
grid2.best_params_
```

```python
y_pred=grid2.best_estimator_.predict(X_test)
```

```python
r2_score(y_test, y_pred)
```
