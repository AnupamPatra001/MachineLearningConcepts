```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
from sklearn.datasets import make_classification
# make_classification() is a function used to generate a synthetic (artificial) 
# classification dataset for testing, learning, and practicing machine learning algorithms.
# Instead of downloading a real dataset, you can create one instantly.
X, y=make_classification(n_samples=1000, n_features=10, n_redundant=5, n_informative=5, random_state=1, n_classes=2)
```

```python
from sklearn.model_selection import train_test_split as tts
X_train, X_test, y_train, y_test = tts(X, y, test_size=0.3, random_state=1)
```

```python
X_train.shape, y_train.shape
```

```python
from sklearn.linear_model import LogisticRegression
model=LogisticRegression()
model
```

```python
model.fit(X_train, y_train)
```

```python
y_pred=model.predict(X_test)
y_pred
```

```python
# evaluation matrics
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
confusion_matrix(y_test, y_pred)
```

```python
accuracy_score(y_test, y_pred)
```

```python
print(classification_report(y_test, y_pred))
```

```python
np.unique(y_train, return_counts=True)
```

```python
from sklearn.metrics import roc_curve, auc
y_pred_proba=model.predict_proba(X_test)[:, 1]
```

```python
# Calculate ROC curve
fpr, tpr, thresholds = roc_curve(y_test, y_pred_proba)
#Calculate AUC score
roc_auc = auc(fpr, tpr)
#Plot ROC curve
plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, color='darkorange', linewidth=2, label='ROC curve (area %0.2f)' % roc_auc)
plt.plot([0, 1], [0, 1], color="navy", lw=2, linestyle='--')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver Operating Characteristic (ROC) Curve')
plt.legend(loc='lower right')
plt.show()
```

```python
# by default prediction class cutoff is 0.5
# if you want to decide own, you have to use precision, recall , accuracy tradeoff
# sensitivity specificity , accuracy tradeoff
```

```python
# imp libraries
from sklearn.metrics import precision_score, recall_score, accuracy_score

# calculate precision, recall, accuracy for different threshold probabilites
thresholds=np.linspace(0, 1, 100)
precisions=[]
recalls=[]
accuracies=[]

for threshold in thresholds:
    y_pred_threshold=(y_pred_proba >= threshold).astype(int)
    precision=precision_score(y_test, y_pred_threshold)
    recall= recall_score(y_test, y_pred_threshold)
    accuracy=accuracy_score(y_test, y_pred_threshold)
    precisions.append(precision)
    recalls.append(recall)
    accuracies.append(accuracy)

# Plot precision, recall, and accuracy against threshold probabilities
plt.figure(figsize=(10, 6))
plt.plot(thresholds, precisions, label='Precision')
plt.plot(thresholds, recalls, label='Recall')
plt.plot(thresholds, accuracies, label='Accuracy')
plt.xlabel('Threshold Probability')
plt.ylabel('Score')
plt.title('Precision, Recall, and Accuracy vs. Threshold Probability')
plt.legend()
plt.grid(True)
plt.show()
    
```

```python
# 0.4 is my optimal threshold
new_pred_levels=np.where(y_pred_proba > 0.4, 1, 0)
new_pred_levels
```

```python
confusion_matrix(y_test, new_pred_levels)
```

```python
accuracy_score(y_test, new_pred_levels)
```

```python
print(classification_report(y_test, new_pred_levels))
```
