```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')
```

```python
from sklearn.datasets import make_classification
X, y=make_classification(n_samples=1000, n_features=10, n_redundant=5, n_informative=5, n_classes=2, random_state=1)
```

```python
from sklearn.model_selection import train_test_split as tts
X_train, X_test, y_train, y_test=tts(X, y, test_size=0.3, random_state=1)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
params={'penalty':['l1','l2','elasticnet'], 'C':[1,2,10, 20, 30, 40]}
```

```python
gclfmodel=GridSearchCV(LogisticRegression(), param_grid=params, cv=5, verbose=2)
gclfmodel
```

```python
gclfmodel.fit(X_train, y_train)
```

```python
gclfmodel.best_params_
```

```python
gclfmodel.best_estimator_
```

```python
gclfmodel.best_score_
```

```python
gclfmodel.best_estimator_.predict(X_test)
```

```python
# or
model=LogisticRegression(C=1, penalty='l2')
model
```

```python
model.fit(X_train, y_train)
```

```python
y_pred=model.predict(X_test)
y_pred
```

```python
# evaluation matrics
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
print(confusion_matrix(y_test, y_pred))
print(accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
rclf=RandomizedSearchCV(LogisticRegression(), param_distributions=params, cv=5, n_iter=10, verbose=2)
rclf
```

```python
rclf.fit(X_train, y_train)
```

```python
rclf.best_estimator_
```

```python
rclf.best_params_
```

```python
rclf.best_score_
```

```python
rclf.best_estimator_.predict(X_test)
```
